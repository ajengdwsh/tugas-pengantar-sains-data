```python
# Import library yang dibutuhkan

import pandas as pd
import numpy as np
from IPython.display import display

print("Library berhasil diimpor!") 
```

```python
Pada bagian ini, saya mengimpor beberapa library Python yang akan digunakan untuk mengolah dan menganalisis dataset. Pandas digunakan untuk membaca dan mengolah data, NumPy digunakan untuk membantu perhitungan numerik, sedangkan display digunakan untuk menampilkan tabel agar hasilnya lebih rapi.
```

```python
# Membaca dataset transaksi penjualan

df = pd.read_csv("paket_a_transaksi_pos_2024_2025.csv")

print("Dataset berhasil dibaca!")
print("Ukuran dataset:", df.shape)

display(df.head())
```

```python
Pada bagian ini, saya membaca dataset transaksi penjualan menggunakan Pandas dengan format CSV. Setelah dataset berhasil dibaca, saya menampilkan ukuran dataset untuk mengetahui jumlah baris dan kolom yang tersedia. Fungsi head() digunakan untuk melihat lima data pertama sebagai gambaran awal isi dataset.
```

```python
# Menampilkan informasi struktur dataset

df.info()
```

```python
Pada bagian ini, saya menggunakan fungsi info() untuk melihat informasi dataset, seperti nama kolom, jumlah data yang terisi, tipe data, dan penggunaan memori. Informasi ini membantu mengetahui kondisi awal dataset sebelum dilakukan proses pembersihan.
```

```python
# Menampilkan statistik deskriptif dataset

display(df.describe(include="all").T)
```

```python
Pada bagian ini, saya menggunakan fungsi describe() untuk melihat gambaran umum dataset. Informasi yang ditampilkan meliputi jumlah data, nilai rata-rata, nilai terkecil, nilai terbesar, serta kategori yang paling sering muncul. Hasil ini digunakan untuk memahami karakteristik data sebelum dilakukan analisis lebih lanjut.
```

```python
# Memeriksa missing value pada setiap kolom

missing = pd.DataFrame({
    "Jumlah Missing": df.isna().sum(),
    "Persentase Missing (%)": (
        df.isna().sum() / len(df) * 100
    ).round(2)
})

display(missing[missing["Jumlah Missing"] > 0])
```

```python
Pada bagian ini, saya memeriksa apakah terdapat nilai kosong atau missing value pada setiap kolom. Hasil pemeriksaan menunjukkan bahwa kolom kode_kasir memiliki 261 data kosong atau sekitar 5,19% dari keseluruhan data.
```

```python
# Mengatasi missing value pada kolom kode_kasir

df["kode_kasir"] = df["kode_kasir"].fillna("TIDAK DIKETAHUI")

print("Pengecekan missing value setelah penanganan:")

display(df.isna().sum().to_frame("Jumlah Missing"))
```

```python
Pada bagian ini, saya mengganti nilai kosong pada kolom kode_kasir dengan tulisan TIDAK DIKETAHUI. Hal ini dilakukan karena kolom tersebut berisi identitas atau kode kasir sehingga tidak tepat jika diisi menggunakan nilai rata-rata. Dengan demikian, data transaksi lainnya tetap dapat digunakan untuk proses analisis.
```

```python
# Mengubah harga_satuan menjadi tipe numerik

harga = (
    df["harga_satuan"]
    .astype("string")
    .str.strip()
    .str.replace(".", "", regex=False)
    .str.replace(",", ".", regex=False)
)

df["harga_satuan"] = pd.to_numeric(
    harga,
    errors="coerce"
)

print("Tipe data harga_satuan:", df["harga_satuan"].dtype)

display(df[["harga_satuan"]].head(10))
```

```python
Pada bagian ini, saya mengubah kolom harga satuan yang awalnya berbentuk teks menjadi angka agar bisa digunakan dalam perhitungan. Format harga yang berbeda diseragamkan terlebih dahulu, sehingga nilai harga dapat dihitung dengan benar saat menentukan total penjualan dan pendapatan bersih.
```

```python
# Mengubah kolom tanggal menjadi datetime

df["tanggal"] = pd.to_datetime(
    df["tanggal"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("Tipe data tanggal:", df["tanggal"].dtype)

display(df[["tanggal"]].head(10))

print(
    "Tanggal yang gagal dikonversi:",
    df["tanggal"].isna().sum()
)
```

```python
Pada bagian ini, saya mengubah kolom tanggal menjadi tipe datetime agar dapat digunakan dalam pengelompokan data berdasarkan tahun. Karena format tanggal dalam dataset berbeda-beda, saya menggunakan format="mixed" supaya berbagai format tanggal dapat dibaca oleh Python.
```

```python
# Memastikan diskon berbentuk numerik

df["diskon_persen"] = pd.to_numeric(
    df["diskon_persen"],
    errors="coerce"
)

print("Daftar nilai diskon:")

display(
    df["diskon_persen"]
    .value_counts(dropna=False)
    .sort_index()
)
```

```python
Pada bagian ini, saya memastikan kolom diskon sudah berbentuk angka agar dapat digunakan dalam perhitungan. Nilai diskon menunjukkan persentase potongan harga, misalnya nilai 10 berarti diskon sebesar 10%, bukan potongan sebesar Rp10.
```

```python
# Membersihkan dan memeriksa status transaksi

df["status"] = (
    df["status"]
    .astype("string")
    .str.strip()
    .str.upper()
)

print("Jumlah transaksi berdasarkan status:")

display(
    df["status"]
    .value_counts()
    .rename_axis("Status")
    .to_frame("Jumlah Transaksi")
)
```

```python
Pada bagian ini, saya memeriksa status dari setiap transaksi untuk mengetahui jumlah transaksi yang selesai, batal, dan retur. Hasil pemeriksaan menunjukkan bahwa terdapat 4.749 transaksi berstatus SELESAI, 191 transaksi BATAL, dan 90 transaksi RETUR. Pemeriksaan ini dilakukan agar nantinya hanya transaksi yang berhasil diselesaikan yang digunakan dalam perhitungan pendapatan.
```

```python
# Menghitung total harga sebelum diskon

df["total_sebelum_diskon"] = (
    df["jumlah"] * df["harga_satuan"]
)

display(
    df[
        [
            "id_transaksi",
            "jumlah",
            "harga_satuan",
            "total_sebelum_diskon"
        ]
    ].head(10)
)
```

```python
Pada bagian ini, saya menghitung total harga setiap transaksi sebelum mendapatkan potongan diskon. Perhitungannya dilakukan dengan mengalikan jumlah barang yang dibeli dengan harga satuan produk. Hasil perhitungan ini nantinya digunakan sebagai dasar untuk menghitung nilai diskon dan pendapatan bersih.
```

```python
# Menghitung nilai potongan diskon

df["nilai_diskon"] = (
    df["total_sebelum_diskon"]
    * df["diskon_persen"] / 100
)

display(
    df[
        [
            "id_transaksi",
            "total_sebelum_diskon",
            "diskon_persen",
            "nilai_diskon"
        ]
    ].head(10)
)
```

```python
Pada bagian ini, menghitung jumlah potongan harga berdasarkan persentase diskon yang diberikan pada setiap transaksi. Nilai diskon diperoleh dengan mengalikan total harga sebelum diskon dengan persentase diskon, kemudian membaginya dengan 100. Dengan demikian, dapat diketahui berapa besar potongan harga yang diterima pada setiap transaksi.
```

```python
Pada bagian ini, menghitung pendapatan bersih dari setiap transaksi setelah dikurangi potongan diskon. Pendapatan bersih diperoleh dari total harga sebelum diskon dikurangi nilai potongan diskon. Dengan perhitungan ini, pendapatan yang diperoleh sudah memperhitungkan diskon yang diberikan kepada pelanggan.

Contohnya, jika pelanggan membeli 4 barang dengan harga Rp13.000 per barang dan mendapatkan diskon 10%, maka:
- Total sebelum diskon = Rp52.000.
- Potongan diskon = Rp5.200.
- Pendapatan bersih = Rp46.800.
```

```python
# Menentukan status transaksi yang digunakan

status_valid = ["SELESAI"]

# Menentukan kolom yang wajib memiliki nilai

kolom_wajib = [
    "tanggal",
    "jumlah",
    "harga_satuan",
    "diskon_persen",
    "status"
]

# Menyaring transaksi yang memenuhi ketentuan

df_valid = df[
    df["status"].isin(status_valid)
    & df[kolom_wajib].notna().all(axis=1)
    & (df["jumlah"] >= 0)
    & (df["harga_satuan"] >= 0)
    & df["diskon_persen"].between(0, 100)
].copy()

print("Total baris dataset:", len(df))
print("Transaksi valid untuk analisis:", len(df_valid))
print("Transaksi yang tidak digunakan:", len(df) - len(df_valid))
```

```python
Pada bagian ini, melakukan penyaringan data transaksi yang akan digunakan dalam analisis. Saya hanya mengambil transaksi dengan status SELESAI serta memastikan tanggal, jumlah barang, harga satuan, dan diskon memiliki nilai yang valid. Transaksi dengan status BATAL dan RETUR tidak dimasukkan dalam perhitungan pendapatan. Selain itu, data dengan jumlah barang atau harga negatif juga tidak digunakan agar hasil perhitungan lebih sesuai.
```

```python
# Mengambil tahun dari tanggal transaksi

df_valid["tahun"] = df_valid["tanggal"].dt.year

display(
    df_valid[
        ["tanggal", "tahun", "pendapatan_bersih"]
    ].head(10)
)
```

```python
Pada bagian ini, mengambil informasi tahun dari setiap tanggal transaksi menggunakan fungsi .dt.year. Tujuannya agar data transaksi dapat dikelompokkan berdasarkan tahun 2024 dan 2025 sehingga total pendapatan dari masing-masing tahun dapat dihitung dan dibandingkan.
```

```python
# Menentukan tahun yang akan dianalisis

tahun_analisis = [2024, 2025]

# Menghitung total pendapatan bersih setiap tahun

hasil_pendapatan = (
    df_valid[df_valid["tahun"].isin(tahun_analisis)]
    .groupby("tahun")["pendapatan_bersih"]
    .sum()
    .reindex(tahun_analisis, fill_value=0)
    .reset_index()
)

# Mengubah nama kolom agar lebih mudah dipahami

hasil_pendapatan.columns = [
    "Tahun",
    "Total Pendapatan Bersih (Rp)"
]

# Membulatkan hasil menjadi bilangan bulat

hasil_pendapatan["Total Pendapatan Bersih (Rp)"] = (
    hasil_pendapatan["Total Pendapatan Bersih (Rp)"]
    .round(0)
    .astype("int64")
)

display(hasil_pendapatan)
```

```python
menjumlahkan seluruh pendapatan bersih dari transaksi yang sudah dinyatakan valid berdasarkan masing-masing tahun. Proses pengelompokan dilakukan menggunakan groupby() sehingga pendapatan tahun 2024 dan 2025 dapat dihitung secara terpisah. Hasilnya ditampilkan dalam bentuk tabel agar lebih mudah dibaca dan dibandingkan.
```

```python
# Mengambil total pendapatan masing-masing tahun

pendapatan_2024 = hasil_pendapatan.loc[
    hasil_pendapatan["Tahun"] == 2024,
    "Total Pendapatan Bersih (Rp)"
].iloc[0]

pendapatan_2025 = hasil_pendapatan.loc[
    hasil_pendapatan["Tahun"] == 2025,
    "Total Pendapatan Bersih (Rp)"
].iloc[0]

# Menghitung selisih pendapatan

selisih = pendapatan_2025 - pendapatan_2024

# Menghitung persentase perubahan

if pendapatan_2024 != 0:
    persentase_perubahan = (
        selisih / pendapatan_2024
    ) * 100
else:
    persentase_perubahan = np.nan

print(f"Pendapatan bersih 2024: Rp{pendapatan_2024:,.0f}")
print(f"Pendapatan bersih 2025: Rp{pendapatan_2025:,.0f}")
print(f"Selisih pendapatan: Rp{selisih:,.0f}")

if pd.notna(persentase_perubahan):
    print(f"Perubahan pendapatan: {persentase_perubahan:.2f}%")
else:
    print("Persentase perubahan tidak dapat dihitung.")
```

```python
membandingkan total pendapatan bersih tahun 2024 dan 2025 untuk mengetahui perubahan pendapatan dari kedua tahun tersebut. Perhitungan dilakukan dengan mencari selisih pendapatan dan persentase perubahannya. Hasil negatif menunjukkan bahwa pendapatan tahun 2025 mengalami penurunan dibandingkan tahun 2024.
```

```python
# Menampilkan kesimpulan hasil analisis

print("KESIMPULAN ANALISIS PENJUALAN")
print("-----------------------------------")

print(
    f"Total pendapatan bersih tahun 2024: "
    f"Rp{pendapatan_2024:,.0f}"
)

print(
    f"Total pendapatan bersih tahun 2025: "
    f"Rp{pendapatan_2025:,.0f}"
)

print(
    f"Selisih pendapatan: Rp{abs(selisih):,.0f}"
)

print(
    f"Persentase perubahan: {abs(persentase_perubahan):.2f}%"
)

if selisih < 0:
    print("Kesimpulan: Pendapatan tahun 2025 mengalami penurunan.")
elif selisih > 0:
    print("Kesimpulan: Pendapatan tahun 2025 mengalami kenaikan.")
else:
    print("Kesimpulan: Pendapatan kedua tahun sama.")
```

```python
rangkuman hasil analisis menggunakan Python. Kesimpulan dibuat berdasarkan perbandingan total pendapatan bersih tahun 2024 dan 2025, sehingga dapat diketahui apakah pendapatan mengalami kenaikan, penurunan, atau tetap.
```
